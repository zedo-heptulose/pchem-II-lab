## Part 1 - Wave Superposition

Maxwell's equations imply that the electric field component $E(x,t)$ of a light wave traveling in one dimension is given by 
$$
E_r(x, t) = \cos\left(\frac{2\pi}{\lambda}x - 2\pi\nu t\right)
$$  
$$
E_\ell(x, t) = \cos\left(\frac{2\pi}{\lambda}x + 2\pi\nu t\right)
$$ 
for light waves moving right and left respectively. Here $\lambda$ is the wavelength and $\nu$ is the frequency. Light waves move at... the speed of light, so $c = \lambda\nu$.

Something interesting happens if we have two overlapping light waves of the same frequency travelling in opposite directions. Using a trigonometric identity, their sum simplifies to:
$$
E(x, t) = E_r + E_\ell = 2 \cos\left(\frac{2\pi}{\lambda}x\right) \cos(2\pi\nu t)
$$
The resulting field has the character of a **standing wave** - it does not travel in either direction. The first cosine factor gives us the spatial structure and points called **nodes** where the field is always zero. The second cosine factor describes how the field oscillates in time.

Qualitatively, how can two traveling waves combine to make a wave that doesn't travel? (Do not use an equation in your answer!)

*Sample answer:*

When two waves traveling in opposite directions meet, they interfere with each other. At some points (nodes), one wave is always going up while the other is going down by the same amount, so they cancel out completely. At other points (antinodes), the waves reinforce each other. The pattern of nodes and antinodes stays fixed in space - only the amplitude oscillates up and down over time. The "traveling" motion of the individual waves cancels out, leaving a stationary pattern that just breathes in and out.

## Part 2 - Wave Quantization

To create a standing light wave, we can place mirrors at a left and right boundary - the light will reflect off the mirrors and interfere with itself. 

At the mirror surfaces, the electric field must be zero (recall from physics that there can be no tangential electric field at a conductor surface). For analogy, imagine a vibrating guitar string fixed at both ends. This gives us two **boundary conditions**:
$$
\phi(0) = 0
$$
$$
\phi(L) = 0
$$

These constraints mean that only waves with zero amplitude at $x=0$ and $x=L$ are allowed. The spatial part of the standing wave, $\phi_n(x)$, must satisfy these boundary conditions, giving us:
$$
\phi_n(x) = \sin\left(\frac{n\pi}{L}x\right)
$$
with associated wavelengths:
$$
\lambda_n = \frac{2L}{n}, \quad n = 1, 2, 3, \dots
$$

Notice that there are now only certain possible waves, and certain possible wavelengths! This is wave **quantization**. If you are thinking "wave quantization" sounds like it has something to do with "quantum mechanics," you would be correct.

## Part 3 - Wave Energies

We've seen that boundary conditions restrict which wavelengths are allowed. But wavelength isn't just a geometric property - it's directly connected to **energy**.

Light comes in discrete packets called **photons**. The energy of a single photon is given by:
$$
E = h\nu = \frac{hc}{\lambda}
$$
where $h$ is Planck's constant, $\nu$ is the frequency, $c$ is the speed of light, and $\lambda$ is the wavelength.

Since our cavity only allows certain wavelengths $\lambda_n = \frac{2L}{n}$, it can only contain photons with certain **quantized energies**:
$$
E_n = \frac{hc}{\lambda_n} = \frac{n \cdot hc}{2L}
$$

The frequency of light is related to wavelength by:
$$
\nu = \frac{c}{\lambda}
$$

This should seem a little strange - light confined to a box can only have a set of discrete energies!

## Part 4 - Standing Waves on Rings and Spheres

Mirrors are not the only way to create boundary conditions. We can also observe quantization when waves propagate on curved surfaces.

Consider a wave traveling around a ring - it must meet itself after going around the full circle. This means the circumference must contain an integer number of wavelengths:
$$
2\pi R = m\lambda, \quad m = 0, 1, 2, 3, \dots
$$

This is another form of quantization - the value of $m$ tells us how many wavelengths fit around the ring.

The same principle extends to waves on a sphere, giving us equations called the **spherical harmonics**, with two parameters $\ell$ and $m$. When you visualize them, ask yourself: "Where have I seen these shapes before?"

#### Part 4B - Spherical Harmonics
`15 points`

- P1.2: Perform basic arithmetic with NumPy arrays

The spherical harmonics $Y_\ell^m(\theta, \phi)$ are the allowed standing wave patterns on a sphere. We will build a function to plot the spherical harmonics ourselves! Most of the heavy lifting is done for you, but you will need to convert from spherical coordinates $(r, \theta, \phi)$ to Cartesian coordinates $(x, y, z)$ so that we can visualize the function.

The conversion formulas are:
$$
x = r \sin(\theta) \cos(\phi)
$$
$$
y = r \sin(\theta) \sin(\phi)
$$
$$
z = r \cos(\theta)
$$

where $r$ is the radial distance from the origin, $\theta$ is the polar angle (from the positive z-axis), ranging from $0$ to $\pi$ and $\phi$ is the azimuthal angle (in the x-y plane from the positive x-axis), ranging from $0$ to $2\pi$.

**Your task:** In the function below, fill in the formulas for `X`, `Y`, and `Z` using the equations above. Use `np.sin()` and `np.cos()` for the trigonometric functions.

```python
from scipy.special import sph_harm
import mpl_toolkits.mplot3d.axes3d as axes3d
import matplotlib.colors as mcolors

def plot_spherical_harmonic(l, m):
    '''
    Plot the real part of the spherical harmonic Y_l^m(θ, φ).
    Adapted from code at https://github.com/DalInar/schrodingers-snake
    '''
    # Create grids of angles
    # Theta: polar angle from 0 to π (north pole to south pole)
    # Phi: azimuthal angle from 0 to 2π (around the equator)
    thetas = np.linspace(0, np.pi, 50)
    phis = np.linspace(0, 2*np.pi, 50)
    
    # Create 2D grids from 1D arrays (meshgrid)
    # This gives us every combination of (theta, phi)
    Theta, Phi = np.meshgrid(thetas, phis) 
    
    # Calculate the spherical harmonic using scipy's sph_harm function
    # Note: scipy uses the convention sph_harm(m, l, phi, theta)
    s_harm = sph_harm(m, l, Phi, Theta)
       
    # Use the absolute value of the real part as the radial distance
    # This determines how far from the origin each point is
    R = np.abs(s_harm.real)
    # we do this to visualize only the real part of the spherical harmonic
    
    # SOLUTION: Convert from spherical coordinates (R, Theta, Phi) to 
    # Cartesian coordinates (X, Y, Z)
    X = R * np.sin(Theta) * np.cos(Phi)
    Y = R * np.sin(Theta) * np.sin(Phi)
    Z = R * np.cos(Theta)

    # Set up the colormap
    cmap = plt.get_cmap('jet')
    norm = mcolors.Normalize(vmin=Z.min(), vmax=Z.max())
    
    # Create the 3D figure
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(1, 1, 1, projection='3d')
    
    # Plot the surface
    ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap=plt.get_cmap('jet'),
                    facecolors=cmap(norm(R)), linewidth=0, antialiased=False, alpha=0.4)

    # Labels and title
    plt.title(r'Real Part of Spherical Harmonics, $Y_l^m(\theta,\phi)$' + r', $l=$' + str(l) + r', $m=$' + str(m))
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$y$')
    ax.set_zlabel(r'$z$')
    
    plt.show()
```
