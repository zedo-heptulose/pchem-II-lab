## The Wave Equation and Standing Waves

In free space, Maxwell's equations imply that the electric field component $E(x,t)$ satisfies the one-dimensional wave equation:

$$ \frac{\partial^2 E(x, t)}{\partial x^2} = \frac{1}{c^2} \frac{\partial^2 E(x, t)}{\partial t^2}$$

The general solutions are traveling waves: $E_r(x, t) = \cos(kx - \omega t)$ (moving right) and $E_\ell(x, t) = \cos(kx + \omega t)$ (moving left), with the dispersion relation $\omega = ck$.

When two counter-propagating waves superpose:
$$E(x, t) = E_r + E_\ell = 2 \cos(kx) \cos(\omega t)$$

This is a **standing wave**: the spatial pattern $\cos(kx)$ has fixed nodes, while $\cos(\omega t)$ causes oscillation in time.

## Quantization from Boundary Conditions

In free space, standing waves are temporary. To sustain them permanently, we need reflections at boundaries.

Imagine two perfectly conducting mirrors at $x=0$ and $x=L$. The boundary conditions require:
$$E(0, t) = 0 \quad \text{and} \quad E(L, t) = 0$$

Only wavelengths that have **nodes at both ends** can persist. The allowed spatial patterns are:
$$\phi_n(x) = \sin \left( \frac{n \pi x}{L} \right)$$

with wavenumbers:
$$k_n = \frac{n\pi}{L}, \quad \lambda_n = \frac{2L}{n}, \quad n = 1, 2, 3, \dots$$

**This is quantization by boundary conditions.** In free space, any wavelength is allowed. Once confined by mirrors, only discrete wavelengths satisfy the boundary conditions.

In real systems - molecules, crystals, or complex cavities - wave equations rarely have closed-form solutions. Instead, we find eigenmodes **numerically** using energy minimization.

## The Rayleigh Quotient

For a 1D cavity with $\phi(0) = \phi(L) = 0$, the standing wave modes satisfy:
$$-\phi''(x) = k^2 \phi(x)$$

We can find these modes by minimizing the **Rayleigh quotient**:
$$R[\phi] = \frac{\int_0^L |\phi'(x)|^2\,dx}{\int_0^L |\phi(x)|^2\,dx}$$

- **Numerator**: measures how rapidly the wave bends ("kinetic energy")
- **Denominator**: normalizes the total amplitude

Minimizing $R[\phi]$ finds the **smoothest wave** satisfying the boundary conditions - this is the ground state!
