# Physical Chemistry of Quantum Dots

Quantum dots are nanoscale semiconductor particles (2-10 nm diameter) that emit light at a wavelength determined by their size. They are used in applications such as QLED displays and biomedical imaging, and show promise for solar energy technologies.

These nanomaterials demonstrate several core concepts in physical chemistry:
- **Molecular Orbital Theory**: How atomic orbitals combine to form molecular orbitals
- **Band Theory**: How discrete MOs become continuous energy bands in solids
- **Fluorescence**: How light absorption and emission relate to electronic structure
- **Quantum Yield**: The efficiency of light emission

We will explore each of these concepts and connect them to quantum dot behavior.

## PART 1: Molecular Orbital Theory (20 points)

You are likely familiar with atomic orbitals from general chemistry—electrons occupy orbitals (1s, 2s, 2p, etc.) according to the Aufbau principle. But how do we describe electronic structure when atoms bond together?

**Molecular orbital (MO) theory** provides the answer. When atomic orbitals overlap:
- **Constructive interference** lowers energy → **bonding MO**
- **Destructive interference** raises energy → **antibonding MO**

Key insight: **N atomic orbitals combine to form N molecular orbitals** (N/2 bonding, N/2 antibonding).

What happens as we combine more and more atoms? Let's find out.

## PART 2: Band Theory and Semiconductors (15 points)

As we extend MO theory to effectively infinite arrays of atoms, discrete molecular orbitals merge into **continuous energy bands**—this is **band theory**.

Solids are classified by their band structure:

| Material Type | Band Gap | Electron Mobility |
|---------------|----------|-------------------|
| **Conductor** | ~0 eV (bands overlap) | High - electrons move freely |
| **Semiconductor** | 0.1-4 eV | Intermediate - can be tuned |
| **Insulator** | >4 eV | Very low - electrons stuck |

**Key concept for quantum dots**: In semiconductors, excited electrons create **excitons** (electron-hole pairs). When a quantum dot is small enough, quantum confinement increases the band gap, shifting emission to higher energy (bluer light).

## PART 3: Fluorescence (15 points)

**Fluorescence** occurs when a material absorbs light and re-emits it at a longer wavelength. The process involves:

1. **Excitation** (femtoseconds): A photon promotes an electron to a higher electronic state
2. **Vibrational relaxation** (picoseconds): The electron loses energy to molecular vibrations
3. **Emission** (nanoseconds): The electron returns to ground state, emitting a photon

Because energy is lost during relaxation, the emitted photon has **lower energy (longer wavelength)** than the absorbed photon. This is called the **Stokes shift**.

The Jablonski diagram on the left shows the energy levels; the spectra on the right show what we measure experimentally.

## PART 4: Predicting Quantum Yield with Machine Learning (25 points)

The **quantum yield** (Φ) measures fluorescence efficiency:

$$\Phi = \frac{\text{photons emitted}}{\text{photons absorbed}} = \frac{k_r}{k_r + k_{nr}}$$

where $k_r$ is the radiative decay rate and $k_{nr}$ is the nonradiative decay rate.

**The challenge**: Quantum dot properties depend on many variables—size, composition, surface chemistry, shell thickness, etc. Testing all combinations experimentally is impractical.

**The solution**: Machine learning can identify patterns and predict which combinations lead to high quantum yield.

We'll use a **Random Forest Classifier** to predict whether a quantum dot will have high (>50%) or low (<50%) quantum yield based on its properties.

### Understanding the Data

We'll use synthetic data representing quantum dots with three features:
- **Diameter** (nm): Size of the quantum dot core
- **Bandgap** (eV): Electronic band gap energy
- **Shell ratio**: Thickness of protective shell relative to core

The target variable is binary: 1 = high quantum yield, 0 = low quantum yield.

```python
# Generate the quantum dot dataset
# (In real research, this would be experimental data)
X, y = generate_quantum_dot_data(n_samples=100, random_state=42)

print(f"Dataset shape: {X.shape}")
print(f"Features: Diameter (nm), Bandgap (eV), Shell Ratio")
print(f"High QY samples: {sum(y):.0f}, Low QY samples: {len(y) - sum(y):.0f}")
```

### TASK 4.1: Train and evaluate a classifier (10 points)

Complete the code below to:
1. Train a Random Forest model on the training data
2. Make predictions on the test data
3. Calculate accuracy

```python
# ============================================
# SUBGOAL: Split data into training and test sets
# ============================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
print(f"Training samples: {len(X_train)}, Test samples: {len(X_test)}")

# ============================================
# SUBGOAL: Create and train the model
# ============================================
model = RandomForestClassifier(n_estimators=10, max_depth=3, random_state=42)

# YOUR CODE HERE: Train the model using .fit() with X_train and y_train
model.fit(X_train, y_train)

# ============================================
# SUBGOAL: Make predictions and evaluate
# ============================================
# YOUR CODE HERE: Make predictions on X_test using .predict()
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nModel accuracy: {accuracy:.1%}")
```

```python
# Plot model predictions
# YOUR CODE HERE: Call plot_model_predictions with X_test, y_test, y_pred, and accuracy
plot_model_predictions(X_test, y_test, y_pred, accuracy)
```
