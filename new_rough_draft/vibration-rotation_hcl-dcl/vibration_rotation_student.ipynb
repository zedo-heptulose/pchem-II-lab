# Part 1: Vibrational Motion — Quantum Harmonic Oscillator

Chemical bonds behave like springs, oscillating around equilibrium bond lengths. These **molecular vibrations** have quantized energy levels, making them fundamental for infrared (IR) spectroscopy and thermodynamics.

In the classical spring model, the restoring force is given by Hooke's Law:
$$F = -kx$$

The corresponding potential energy is:
$$V(x) = \frac{1}{2}kx^2$$

The quantum Hamiltonian becomes:
$$\hat{H}\psi(x) = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}\psi(x) + \frac{1}{2}m\omega^2x^2\psi(x)$$

where $\omega$ is the angular frequency. We use units where $\hbar = 1$ and $m = 1$ to simplify calculations.

## Task: Explore Quantum Harmonic Oscillator Wavefunctions

The QHO eigenfunctions have the form:
$$\psi_n(x) = H_n(\sqrt{\omega}x) \cdot e^{-\omega x^2/2}$$

where $H_n$ are Hermite polynomials. The first few are:
- $n=0$: $H_0 = 1$
- $n=1$: $H_1 = 2x$
- $n=2$: $H_2 = 4x^2 - 2$

# Part 2: Rotational Motion — Rigid Rotor

The **rigid rotor** model describes rotational motion, assuming atoms remain at fixed distances. Rotation occurs in three dimensions, so we use **spherical coordinates**: $r$, $\theta$, $\phi$.

The wavefunctions for a rigid rotor are the **spherical harmonics**:
$$Y_J^m(\theta, \phi)$$

where:
- $J$ = rotational quantum number (0, 1, 2, ...)
- $m$ = magnetic quantum number ($-J \leq m \leq J$)

The energy levels depend only on $J$:
$$E_J = \frac{\hbar^2}{2I} J(J+1)$$

where $I$ is the moment of inertia.

# Part 3: Applications — Reaction Thermodynamics

Translational, vibrational, and rotational motion are essential for predicting **reaction thermodynamics**. A single-point energy gives only electronic energy, but thermodynamic properties like $\Delta G$, $\Delta H$, and $\Delta S$ require vibrational frequencies.

The standard workflow is:
1. **Geometry optimization** — Find the equilibrium structure
2. **Frequency calculation** — Compute vibrational modes at equilibrium

Quantum chemistry programs use the **Rigid Rotor Harmonic Oscillator (RRHO) approximation**, treating translation, vibration, and rotation independently.

## The Diels-Alder Reaction

We'll compute thermodynamics for the Diels-Alder reaction:

**ethene + butadiene → cyclohexene**

### Given Functions

Run these cells to define helper functions for molecular specification.

```python
def xyz_from_smiles(smiles_string):
    """Convert SMILES to 3D XYZ coordinates."""
    rdkit_molecule = Chem.MolFromSmiles(smiles_string)
    rdkit_molecule = Chem.AddHs(rdkit_molecule)
    result = AllChem.EmbedMolecule(rdkit_molecule)
    if result != 0:
        raise ValueError("Embedding failed for the molecule")
    result = AllChem.MMFFOptimizeMolecule(rdkit_molecule)
    if result != 0:
        raise ValueError("Optimization failed for the molecule")
    xyz = Chem.MolToXYZBlock(rdkit_molecule)
    return xyz
```

```python
def show_molecule(smiles_string):
    """Display a 3D view of a molecule from SMILES."""
    xyz = xyz_from_smiles(smiles_string)
    view = py3Dmol.view(width=200, height=200)
    view.addModel(xyz, 'xyz')
    view.setStyle({'sphere': {'radius': 0.3}, 'stick': {'radius': 0.2}})
    view.setStyle({'element': 'H'}, {'sphere': {'radius': 0.3, 'color': 'white'}})
    view.zoomTo()
    view.show()
```

```python
def create_psi4_molecule(smiles_string):
    """Convert SMILES to a Psi4 molecule object."""
    xyz_block = xyz_from_smiles(smiles_string)
    xyz_lines = xyz_block.split('\n')
    psi_coords = "\n".join(["0 1"] + xyz_lines[2:])
    psi4_molecule = psi4.geometry(psi_coords)
    return psi4_molecule
```

In real IR spectra, peaks are broadened due to:
- Anharmonicity
- Thermal population of excited states
- Solvent interactions
