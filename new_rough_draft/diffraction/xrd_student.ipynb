# Part 2: From Double Slit to Diffraction

The double slit experiment demonstrates wave interference with light. Constructive interference occurs when:
$$d \cdot \sin(\theta) = n\lambda$$

where $d$ is slit spacing, $\theta$ is the angle, $n$ is an integer, and $\lambda$ is wavelength.

**Key insight:** Intensity is proportional to amplitude squared: $I \propto A^2$

This is analogous to quantum mechanics where probability density equals wavefunction amplitude squared: $|\psi|^2$.

# Part 3: Crystal Lattices and Bragg's Law

Crystals exhibit **translational symmetry**: the structure repeats in three directions. We can describe a crystal entirely by its **unit cell** and **cell parameters** ($a$, $b$, $c$ for lengths; $\alpha$, $\beta$, $\gamma$ for angles).

X-ray diffraction occurs when X-rays scatter off atoms in the crystal. Constructive interference happens when **Bragg's Law** is satisfied:

$$n\lambda = 2d \cdot \sin(\theta)$$

where $d$ is the spacing between parallel planes of atoms.

## Miller Indices (hkl)

Different sets of parallel planes through a crystal are labeled by **Miller indices** $(h, k, l)$. For a cubic lattice, the spacing between planes is:

$$d = \frac{a}{\sqrt{h^2 + k^2 + l^2}}$$

where $a$ is the unit cell parameter.

# Part 4: Analyzing XRD Spectra

In powder XRD, the spectrum shows intensity vs. $2\theta$. We observe **selection rules** based on crystal symmetry:

| Lattice Type | Selection Rule |
|--------------|----------------|
| Simple cubic | All $(h,k,l)$ allowed |
| Face-centered cubic (FCC) | $h,k,l$ all even OR all odd |
| Body-centered cubic (BCC) | $h+k+l$ must be even |

By indexing the peaks with Miller indices, we can determine both the lattice spacing AND the crystal type.
