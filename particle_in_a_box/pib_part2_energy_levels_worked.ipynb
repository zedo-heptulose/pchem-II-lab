## Part 2 — Energy Levels of the Particle in a Box

In Part 1, we used numerical minimization to find the ground-state wavefunction — the lowest-energy solution. But the ground state is not the only solution. The particle-in-a-box has an infinite number of allowed states, each with a different number of nodes and a different energy.

In this part, you will discover these excited states numerically, explore how energy depends on the quantum number $n$ and the box length $L$, and empirically determine the relationship $E(n, L)$.

```python
# --- Given: plotting utility (same as Part 1) ---
def plot_wavefunction(x, y, label=None):
    """Plot a wavefunction and display its energy."""
    E = compute_energy(y, x)
    plt.plot(x, y, label=label + f" (E={E:.3f})")
    plt.axhline(0, color='grey', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=1, linestyle='--', alpha=0.4)
    plt.axvline(x[-1], color='black', linewidth=1, linestyle='--', alpha=0.4)
    plt.xlabel('x (a.u.)')
    plt.ylabel(r'$\psi(x)$')
    plt.legend()
```

#### Part A — Excited states

In Part 1, you saw that the optimizer finds the ground-state wavefunction by minimizing the energy. But the ground state is not the only eigenfunction — the functions $\psi_n(x) = \sin(n\pi x / L)$ for $n = 1, 2, 3, \ldots$ are all eigenfunctions of the Hamiltonian.

To confirm this, construct $\psi_n$ for $n = 1, 2, 3$ at a fixed box length, optimize each one, and verify that the optimizer does not change them. Use `plot_wavefunction` to compare before and after.

```python
# n = 1 (worked example)
L = 4.0
x = np.linspace(0, L, 200)

psi_1 = np.sin(1 * np.pi * x / L)
result_1 = minimize(compute_energy, psi_1, args=(x,), method='CG')

plot_wavefunction(x, psi_1, label='n=1 (before)')
plot_wavefunction(x, result_1.x, label='n=1 (after)')
plt.title('n = 1: Before and After Optimization')
plt.show()
```

```python
# n = 2
psi_2 = np.sin(2 * np.pi * x / L)
result_2 = minimize(compute_energy, psi_2, args=(x,), method='CG')

plot_wavefunction(x, psi_2, label='n=2 (before)')
plot_wavefunction(x, result_2.x, label='n=2 (after)')
plt.title('n = 2: Before and After Optimization')
plt.show()

# n = 3
psi_3 = np.sin(3 * np.pi * x / L)
result_3 = minimize(compute_energy, psi_3, args=(x,), method='CG')

plot_wavefunction(x, psi_3, label='n=3 (before)')
plot_wavefunction(x, result_3.x, label='n=3 (after)')
plt.title('n = 3: Before and After Optimization')
plt.show()
```

#### Part C — Discovering the energy formula

You've seen how $E$ depends on $n$ and $L$ individually. Now let's put it all together. The code below computes $E$ for all combinations of $n$ and $L$, then plots the data alongside a model prediction.

**Your task:** The model function `energy_model` does not fit the data. Modify it until the surface matches the data points. Add comments to the provided code explaining what each section does.

```python
# --- Provided: compute energies for all (n, L) combinations ---
n_values = [1, 2, 3, 4, 5, 6]
L_values = [4, 6, 8, 10, 12]

n_list, L_list, E_list = [], [], []

for n in n_values:
    for L in L_values:
        x = np.linspace(0, L, 500)
        psi = np.sin(n * np.pi * x / L)
        E = compute_energy(psi, x)
        n_list.append(n)
        L_list.append(L)
        E_list.append(E)

n_arr = np.array(n_list)
L_arr = np.array(L_list)
E_arr = np.array(E_list)
```

```python
# Modify this function until the model surface matches the data points!
# curve_fit will optimize a, b, c for you — you just need the right functional form.
def energy_model(nL, a, b, c):
    n, L = nL
    return a * n**b / L**c

from scipy.optimize import curve_fit
popt, pcov = curve_fit(energy_model, (n_arr, L_arr), E_arr, p0=[1.0, 2.0, 2.0])
a_fit, b_fit, c_fit = popt
print(f"Fitted parameters: a = {a_fit:.4f}, b = {b_fit:.4f}, c = {c_fit:.4f}")
```

```python
# --- Provided: plot data points and model surface ---
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Numerical data as scatter points
ax.scatter(L_arr, n_arr, E_arr, c='red', s=50, label='Numerical data')

# Model surface
n_grid, L_grid = np.meshgrid(np.linspace(1, 6, 30), np.linspace(4, 12, 30))
E_model = energy_model((n_grid, L_grid), *popt)
ax.plot_surface(L_grid, n_grid, E_model, alpha=0.3, cmap='viridis')

ax.set_xlabel('Box length L (a.u.)')
ax.set_ylabel('Quantum number n')
ax.set_zlabel('Energy (a.u.)')
ax.set_title('PIB Energy: Data vs. Model')
ax.legend()
plt.tight_layout()
plt.show()
```
