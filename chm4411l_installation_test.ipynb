## 2. SciPy: Scientific Functions
Used for curve fitting, Fourier transforms, and special functions.

```python
from scipy.optimize import curve_fit
from scipy.special import sph_harm
from scipy.fft import fft, ifft, fftfreq
import scipy

# Test curve fitting
def linear(x, m, b):
    return m * x + b

x_data = np.array([1, 2, 3, 4, 5])
y_data = np.array([2.1, 4.0, 5.9, 8.1, 9.9])
popt, _ = curve_fit(linear, x_data, y_data)

# Test FFT
signal = np.sin(2 * np.pi * 5 * np.linspace(0, 1, 100))
spectrum = fft(signal)

# Test spherical harmonics
Y_00 = sph_harm(0, 0, 0, 0)

print(f"SciPy version: {scipy.__version__}")
print(f"\nCurve fit test: slope = {popt[0]:.2f}, intercept = {popt[1]:.2f}")
print(f"FFT test: spectrum has {len(spectrum)} points")
print(f"Spherical harmonic Y(0,0) = {Y_00:.4f}")
print("\nSciPy OK!")
```

## 4. IPyWidgets: Interactive Visualizations
Used for creating interactive controls in notebooks.

```python
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, IntSlider
from IPython.display import display, clear_output

print(f"IPyWidgets version: {widgets.__version__}")
print("\nIPyWidgets OK!")
```

```python
# Test interactive widget
def plot_wave(frequency=1.0, amplitude=1.0):
    """Plot a simple sine wave with adjustable parameters."""
    clear_output(wait=True)
    x = np.linspace(0, 2*np.pi, 200)
    y = amplitude * np.sin(frequency * x)
    
    plt.figure(figsize=(8, 3))
    plt.plot(x, y, 'b-', linewidth=2)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'y = {amplitude:.1f} sin({frequency:.1f}x)')
    plt.ylim(-3, 3)
    plt.grid(True, alpha=0.3)
    plt.show()

widgets.interact(
    plot_wave,
    frequency=(0.5, 5.0, 0.5),
    amplitude=(0.5, 2.5, 0.5)
);
```

## 6. RDKit & Py3Dmol: Molecular Structures
Used for working with molecular structures and 3D visualization.

```python
import rdkit
from rdkit import Chem
from rdkit.Chem import AllChem
import py3Dmol

def xyz_from_smiles(smiles_string):
    """Convert a SMILES string to XYZ coordinates."""
    rdkit_molecule = Chem.MolFromSmiles(smiles_string)
    rdkit_molecule = Chem.AddHs(rdkit_molecule)
    result = AllChem.EmbedMolecule(rdkit_molecule, randomSeed=42)
    if result != 0:
        raise ValueError("Embedding failed for the molecule")
    result = AllChem.MMFFOptimizeMolecule(rdkit_molecule)
    if result != 0:
        raise ValueError("Optimization failed for the molecule")
    return Chem.MolToXYZBlock(rdkit_molecule)

print(f"RDKit version: {rdkit.__version__}")
print("\nRDKit OK!")
```

```python
# Test 3D molecular visualization with water molecule
water_smiles = "O"
xyz = xyz_from_smiles(water_smiles)

view = py3Dmol.view(width=400, height=300)
view.addModel(xyz, 'xyz')
view.setStyle({'sphere': {'radius': 0.3}, 'stick': {'radius': 0.2}})
view.zoomTo()
view.show()

print("\nPy3Dmol OK!")
```

## 7. Psi4: Quantum Chemistry Calculations
Used for electronic structure calculations.

**Note:** This test performs an actual quantum chemistry calculation and may take 10-30 seconds.

```python
import psi4

# Configure Psi4
psi4.set_output_file('psi4_test_output.dat', False)
psi4.set_memory('500 MB')

print(f"Psi4 version: {psi4.__version__}")
print("\nPsi4 import OK!")
```

```python
# Test Psi4 calculation with water molecule
water_geometry = psi4.geometry("""
0 1
O   0.000000   0.000000   0.117369
H  -0.756950   0.000000  -0.469476
H   0.756950   0.000000  -0.469476
""")

# Run a simple Hartree-Fock calculation
energy, wfn = psi4.energy('SCF/STO-3G', return_wfn=True, molecule=water_geometry)

print(f"Water energy: {energy:.6f} Hartree")
print(f"Number of orbitals: {wfn.nmo()}")
print("\nPsi4 calculation OK!")
```

Used for visualizing molecular orbitals from cube files.

```python
import fortecubeview
import os
import shutil

# Create cube files for water orbitals
os.makedirs('test_cubes', exist_ok=True)

psi4.set_options({
    'CUBEPROP_TASKS': ['orbitals'],
    'CUBEPROP_FILEPATH': 'test_cubes',
    'CUBEPROP_ORBITALS': [1, 2, 3, 4, 5, 6, 7]  # All orbitals for water
})
psi4.cubeprop(wfn)

# Display the orbitals
view = fortecubeview.plot('test_cubes', colorscheme='emory')

print("\nFortecubeview OK!")
```
