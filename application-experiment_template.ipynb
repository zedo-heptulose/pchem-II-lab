**Chemistry Learning Objectives:**
> Choosing chemistry learning objectives is easy. it's more or less just, "what are the four sections of the lab?"  
> The flow of these (Photoelectric, Quantum Dots, Ruby Laser) is pretty much the same, and it looks like:
> - students are given a lot of raw spectroscopic data, too much to process manually
> - In part one, the students load and plot one or a few representative spectra and extract a constant from the spectrum
> - In part two, the students automate carrying out this process for ALL the data
> - In part three, the students explore the data in one way ... (ex. photoelectric, interpreting shapes of data, dependence of V0 on intensity vs frequency and what this means) 
> - and in part four, the students do something else with it. (ex. photoelectric, extracting Planck's constant and the work function from V0 vs frequency)

- C#.1: for Part 1 (relate to loading and interpreting a spectrum, maybe to extract a constant)
- C#.2: for Part 2 (relate to processing all the data)
- C#.3: for Part 3 (doing something with the data)
- C#.4: for Part 4 (doing something else with the data)

**Programming Learning Objectives:**
> These are a little harder, but in general two new syntax/library techniques and connect to a mental model. Leads to transfer which we want.

> specific things to hit for mental models- 
> - abstraction (seeing the purpose of blocks of code, instead of line-for-line), 
> - decomposition (breaking down a problem), 
> - metacognition (thinking about how you approach a problem, planning and keeping track of progress)

- P#.1: A mental model (functions as physical models, for example)
- P#.2: Syntax or a library
- P#.3: Syntax or a library

`10 points`

> Reflection questions aren't about the scientific content of the lab per se.
> These are more to make the students think about their problem solving process, or more philosophical patterns in chemistry.

> Idea here is to build transferable mental models related to what they learned. We get a twofer for metacognition and feedback with the last question.

a) How did [approach used in this lab] compare to the approach taken in lecture? Which do you prefer?

b) [question about overarching chemistry or programming mental model] (ex. how would you evaluate models, how do you interpret functions, etc)

c) When completing this lab, what did you feel most and least confident about and why? What would have helped you feel more confident?
